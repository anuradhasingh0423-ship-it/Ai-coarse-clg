```python
df.loc[df['diagnosis']=='M','benign_0_mal_1']=1
df.loc[df['diagnosis']=='B','benign_0_mal_1']=0
```

```python
#dropping the non related cols
df=df.drop(columns=['id','diagnosis'])
```

```python
#seperating independent and dependent variables
x=df.drop(['benign_0_mal_1'],axis=1).values
y=df['benign_0_mal_1'].values
```

```python
#Train test Split
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=101)
```

```python
y_test.shape
```

```python
X_train.shape,X_test.shape
```

```python
#Feature Scaling
from sklearn.preprocessing import MinMaxScaler
scaler=MinMaxScaler()
X_train=scaler.fit_transform(X_train)
X_test=scaler.fit_transform(X_test)
```

```python
#X_train_df=pd.DataFrame(x_train)
#_train_df.head()
```

```python
!pip install tensorflow
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense,InputLayer,Dropout
```

```python
#creating the model
model1=Sequential()
model1.add(InputLayer(input_shape=(31,)))#Input Layer with 26 input features
model1.add(Dense(26,activation='relu')) #first hidden layer
model1.add(Dense(15,activation='relu'))#second hidden layer
model1.add(Dense(1,activation='sigmoid'))#output layer
model1.summary()
```

```python
#for binary classification loss is binary_crossentory
model1.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])
```

```python
model1.fit(x=X_train,y=y_train,epochs=100,validation_data=(X_test,y_test))
```

```python
model_loss=pd.DataFrame(model1.history.history)
model_loss.plot()
```
