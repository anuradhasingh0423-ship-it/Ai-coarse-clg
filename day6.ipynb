```python
# take 2 numbers from user and print the sum of their squares

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

result = a**2 + b**2
print("Sum of squares:", result)
```

```python
# take a word from user and print its reverse
word = input("Enter a word: ")
print("Reversed word:", word[::-1])
```

```python
# a=[31,32,33,34,35,36] , print the first 3 elements of list
a = [31, 32, 33, 34, 35, 36]

# first 3 elements
print(a[:3])

# add 37
a.append(37)
print(a)

# remove 3rd element (index = 2)
a.pop(2)
print(a)
```

## Heatmap

it is used for correlation betn cols

correlarion varies from -1 to 1

nearer to 0 means no correlation 

nearer to -1 or 1 means strong correlation

nearer to -1 means inverse relation 

nearer to 1 means direct relation

nearer to 0 means no relation


```python
df.columns
```

```python
sns.heatmap(df[['Pulse','Calories','Maxpulse']].corr(), annot = True)
```

```python
sns.heatmap(df[['Pulse','Calories','Maxpulse','Duration']].corr(), annot = True)
```
