```python
''' user have 3 flowers initially as he takes a step 
the flowers get doubled, take number of steps taken from user
and give the number of flowers user have at last'''

s = int(input("Enter number of steps: "))

initial_flowers = 3
final_flowers = initial_flowers *( 2 ** s)

print("Number of flowers at last:", final_flowers)
```

```python
# que1
# create a program that takes the user's height and weight as input and calculates their Body Mass index(BMI)
# print the result in the format "your BMI is" (BMI)

height = int(input("Enter your height in meters: "))
weight = int(input("Enter your weight in kilograms: "))

bmi = weight / (height ** 2)

print("Your BMI is", bmi)
```

```python
# isnumeric()


a = input('enter something: ')
print(a.isnumeric())
```

```python
text = input("Enter a string: ")

print("Lowercase:", text.lower())
print("Uppercase:", text.upper())
print("Title case:", text.title())
print("Capitalized:", text.capitalize())
print("Length of string:", len(text))
print("Stripped string:", text.strip())
print("Number of 'a':", text.count('a'))
print("Replace a with @:", text.replace('a', '@'))
print("Starts with 'a':", text.startswith('a'))
print("Ends with 'z':", text.endswith('z'))
```
