### NLP
it is a branch of Artificial Intelligence that enables computers to understand, interpret, and generate human language (text or speech).

It sits at the intersection of AI, Machine Learning, and Linguistics, helping machines work with language the way humans do.

## 1️⃣ Natural Language Understanding (NLU)

 Helps the computer understand human language.

🔹 What NLU does:

Understands meaning, context, and intent

Converts text into structured data

🔹 Key NLU components:

Lexical Analysis – word-level processing

Syntactic Analysis (Parsing) – grammar & sentence structure

Semantic Analysis – meaning of words & sentences

Discourse Integration – meaning across sentences

Pragmatic Analysis – meaning based on real-world context

 Example:
“Book a flight to Delhi”
→ Intent: Flight booking
→ Entity: Delhi


## 2️⃣ Natural Language Generation (NLG)

 Helps the computer generate human-like language.

🔹 What NLG does:

Converts data into readable text

Produces responses, summaries, reports

🔹 Key NLG components:

Content Determination – what to say

Sentence Planning – how to say it

Text Realization – grammar & fluency

 Example:
Data → “Your flight to Delhi is confirmed.”

## **Stopwords in NLP**

```python
from nltk.corpus import stopwords

nltk.download('stopwords')
```

```python
len(stopwords.words('english'))
```

```python
print(stopwords.fileids())
```

```python
text = """He determined to drop his litigation with the monastry, and relinguish his claims to the wood-cuting and
          fishery rihgts at once. He was the more ready to do this becuase the rights had become much less valuable, and he had
          indeed the vaguest idea where the wood and river in question were. """

```

```python
# set of stop words
stop_words = set(stopwords.words('english'))

# tokens of words
word_tokens = word_tokenize(text)
filtered_sentence = [] # list of storing words other than stop words

for w in word_tokens:
  if w not in stop_words:
    filtered_sentence.append(w)
```

```python
print("\n Original Sentence \n")
print(" ".join(word_tokens))

print("\n Filtered Sentence \n")
print(" ".join(filtered_sentence))
```

## **SpaCy**

```python
import spacy

# !python -m spacy download en_core_web_sm

from collections import Counter
from string import punctuation
```

```python
nlp = spacy.load('en_core_web_sm')
```

```python
def get_hotwords(text):
  result = []
  pos_tag = ['PROPN', 'ADJ', 'NOUN']
  doc = nlp(text.lower())
  for token in doc:
    if(token.text in nlp.Defaults.stop_words or token.text in punctuation):

      continue
    if(token.pos_ in pos_tag):
      result.append(token.text)
  return result
```

```python
new_text = """
           When it comes to evaluating the performance of keyword extractors, you can use some of the standard metrics in machine learning: accuracy, precision, recall, and F1 score. However, these metrics don’t reflect partial matches. they only consider the perfect match between an extracted segment and the correct prediction for that tag.
           Fortunately, there are some other metrics capable of capturing partial matches. An example of this is ROUGE.
           """
```

```python
output = set(get_hotwords(new_text))
most_common_list = Counter(output).most_common(10)

for item in most_common_list:
  print(item[0])
```

## **Porter's Stemmer**

```python
from nltk.stem import PorterStemmer
ps = PorterStemmer()

print(ps.stem('stemming'))
```

```python
sentence =  "It is good to be a pythoner who is pythonly pythoning with python"
words = nltk.word_tokenize(sentence)
for word in words:
  print(ps.stem(word))
```

## **Lancaster Stemmer**

```python
from nltk.stem import SnowballStemmer, LancasterStemmer
sentence = "It is good to be a pythoner who is pythonly pythoning with python"
words = nltk.word_tokenize(sentence)
for word in words:
  lanc = LancasterStemmer()
  print(lanc.stem(word))
```
