### Matplotlib

MATPLOTLIB

Types Of Chart

Pie chart : to visualize the part perpostion and percentage

Line chart : to view the trends ups and downs

Histogram : it is used to visualize distribution of col

Box plot : it is used for outliers

Scatter plot : it is used to understand relation bet 2 numeric cols

Heatmap : to see correlation bet multiple cols

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([0,6])
y = np.array([0,250])

plt.plot(x,y)
plt.show()
```

```python
y = np.array([3, 8, 1, 10, 5, 7 ])
plt.plot(y)
plt.show()
```

```python
y = np.array([3, 8, 1, 10, 5, 7])

plt.plot(y, linestyle = 'dotted')  # represnt dotted lines
plt.show()
```

```python
y = np.array([3, 8, 1, 10, 5, 7])

plt.plot(y, 'o:r')  # represnt particular points
plt.show()
```

```python
# plot 1:
x = np.array([0, 1, 2, 3])
y = np.array([3, 8, 1, 10])

plt.subplot(1, 2, 1)    # plt.subplot(size, position)
plt.plot(x,y)

# plot 2:
x = np.array([0, 1, 2, 3])
y = np.array([10, 20, 30, 40])

plt.subplot(1, 2, 2)    # plt.subplot(size, position)
plt.plot(x,y)

plt.show()
```

```python
## inverse x inse and y decre
x = np.array([5, 7, 8, 7, 2, 17, 2, 9, 4,11, 12, 9, 6])
y = np.array([99,86,87,88,111,86,103,87,94,78,77,85,86])
plt.scatter(x,y)
plt.show()
```

```python
x = np.array(["A", "B", "C", "D"])
y = np.array([3, 8, 1, 10])
plt.bar(x,y)
plt.show()
```

```python
x = np.array(["A", "B", "C", "D"])
y = np.array([3, 8, 1, 10])
plt.barh(x,y)
plt.show()
```
