```python
# take angle from user and classify it as acute, right or obtuse angle
angle = int(input("Enter the angle: "))

if angle < 90:
    print("Acute angle")
elif angle == 90:
    print("Right angle")
elif angle < 180:
    print("Obtuse angle")
else:
    print("Not a valid angle")
```

```python
# take 3 sides of a triange and classify it as equailateral, isoceles or scalene triange
a = int(input('Enter side1:'))
b = int(input('Enter side2:'))
c = int(input('Enter side3:'))

if(a==b==c):
   print('Equilateral triange')
elif (a==b or b==c or a==c ):
   print('isoceles triange')
else:
   print('scalene')
```

```python
# take 3 sides of a triange and classify it as equailateral, isoceles or scalene triange
a = int(input('Enter side1: '))
b = int(input('Enter side2: '))
c = int(input('Enter side3: '))

# check if triangle is valid
if a + b > c and a + c > b and b + c > a:
    if a == b == c:
        print('Equilateral triangle')
    elif a == b or b == c or a == c:
        print('Isosceles triangle')
    else:
        print('Scalene triangle')
else:
    print('Not a valid triangle')


```

```python
# take age and gender and check retirement eligibility, for males the age should be more than 60 and for females age should
# should be greater than 55 then they are eligible for retirement

age = int(input("Enter your age: "))
gender = input("Enter your gender (male/female): ").lower()

if gender == "male" and age > 60:
    print("Eligible for retirement")
elif gender == "female" and age > 55:
    print("Eligible for retirement")
else:
    print("Not eligible for retirement")
```

```python
# create a calculator that calcultate addn,subtractn,multi,and division betweeen 2 numbers

num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))

print("1. Addition")
print("2. Subtraction")
print("3. Multiplication")
print("4. Division")

choice = int(input("Enter your choice (1/2/3/4): "))

if choice == 1:
    print("Result of Addition:", num1 + num2)
elif choice == 2:
    print("Result of substraction:", num1 - num2)
elif choice == 3:
    print("Result of multiplication:", num1 * num2)
elif choice == 4:
    if num2 != 0:
        print("Result of division:", num1 / num2)
    else:
        print("Cannot divide by zero")
else:
    print("Invalid choice")
```
