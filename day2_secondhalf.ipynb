```python
j = [12,23,34,45,56,67]

# slicing in list
print(j[4])

# extract 34
print(j[1])

# extract [23,34,45]
print(j[1:4])

# reverse the list
print(j[:-1])

# extract the 2nd last element
print(j[-2])

# extract [23,45]
print(j[1:4:2])

# print the total number of elemts
print(len(j))
```

```python
j = [12, 23, 34, 45, 56, 67]

# slicing in list (5th element: index 4)
print(j[4])          # 56

# extract 34
print(j[2])          # 34  (index of 34 is 2, not 1)

# extract [23, 34, 45]
print(j[1:4])        # [23, 34, 45]

# reverse the list
print(j[::-1])       # [67, 56, 45, 34, 23, 12] 
# your j[:-1] was incorrect — that excludes last element instead of reversing

# extract the 2nd last element
print(j[-2])         # 56

# extract [23, 45]
print(j[1:5:2])      # [23, 45] (stop at index 5 so it includes 45)

# print total number of elements
print(len(j))        # 6
```

```python
# adding elements
# append >> add to last

a = [1,2,3,4]
# add 5 to list

a.append(6)
print(a)
```

```python
# insert >> to add element at desired index

d = [12,23,34,45]
# add 20 at 2nd place

d.insert(1,20) # list.insert(index, element)
print(d)


# add 30 at 3rd index
d.insert (3,3)
print(d)
```

```python
# extend >> to add multiple elements
a = [12,23,34,45]

# add 50 and 60 to list
a.extend([50, 60])
print(a)

# add 80,90 and 100 to list
a.extend([80,90,100])
print(a)
```

```python
a = [12,23,34,45]
# removing elements
a.remove(34)
print(a)

a.remove(12)
print(a)
```

```python
# aggregation 
# min, max, sum, count

s = [12, 23, 34, 45, 56, 56, 56, 56]

print(min(s))          # minimum value
print(max(s))          # maximum value
print(s.count(56))     # frequency of 56

# total addition of all elements
print(sum(s))

# print the frequency of 34
print(s.count(34))

# print avg of all elements
avg = sum(s) / len(s)
print(avg)
```
