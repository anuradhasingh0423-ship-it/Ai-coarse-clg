```python
# create a function that gives the negative index of element of a list
def negative_index(lst, element):
    if element in lst:
        return lst.index(element) - len(lst)
    else:
        return None
negative_index([1,2,2,3,4,5,6,7],6)
```

```python
# create a function that takes afternoon time in 24hrs format
# and give that time in 12hrs

def convert_to_12hr(time24):
    # Split hours and minutes
    hours, minutes = map(int, time24.split(':'))
    
    # Convert
    if hours == 12:
        period = "PM"
    else:
        period = "PM"
        hours = hours - 12
    
    # Format result
    return f"{hours}:{minutes:02d} {period}"

# Example
time = input("Enter afternoon time (24hr format): ")
print(convert_to_12hr(time))
```
