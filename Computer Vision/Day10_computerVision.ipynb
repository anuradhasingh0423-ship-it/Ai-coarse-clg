```python
img_BGR = cv2.imread('/content/lovers.jpeg')
img_RGB = cv2.cvtColor(img_BGR, cv2.COLOR_BGR2RGB)
plt.imshow(img_RGB)
```

```python
# convert to grayscale
img_gray = cv2.cvtColor(img_BGR, cv2.COLOR_BGR2GRAY)
plt.imshow(img_gray, cmap='gray')
#
```

```python
# blur the img
img_blur = cv2.GaussianBlur(img_gray, (3,3), 0)
plt.imshow(img_blur, cmap='gray')
```

## **Sobel Edge Detection**


Sobel detects edges by measuring intensity change (gradient) in an image.

Big change in pixel values → edge

Small change → flat region

```python
sobelx = cv2.Sobel(src=img_blur, ddepth=cv2.CV_64F, dx=1,dy=0, ksize=5) # Sobel Edge Detection on the X axis
sobely = cv2.Sobel(src=img_blur, ddepth=cv2.CV_64F, dx=0,dy=1, ksize=5) # Sobel Edge Detection on the Y axis
sobelxy = cv2.Sobel(src=img_blur, ddepth=cv2.CV_64F, dx=1,dy=1, ksize=5) # Combined X and Y Sobel Edge Detection
```

```python
# show the image
plt.figure(figsize=[18,5])
plt.subplot(141);
plt.imshow(sobelx, cmap='gray');
plt.title("Sobel X")


plt.subplot(142);
plt.imshow(sobely, cmap='gray');
plt.title("Sobel Y")


plt.subplot(143);
plt.imshow(sobelxy, cmap='gray');
plt.title("Sobel XY")
```

## **Canny**

it is a multi-stage edge detector that:

Reduces noise

Finds gradients

Keeps only strong edges

Removes false edges

Result → thin, clean, continuous edges

```python
canny_edges = cv2.Canny(image=img_blur, threshold1=130, threshold2=200)
plt.imshow(canny_edges, cmap='gray')
plt.title("Canny Edge Detection")

# any gradient below 130 -> rejected
# above 200 -> definitely an edge
```
