```python
import cv2
from google.colab.patches import cv2_imshow

image_filepath = "/content/aish.jpg"

# Load the image in color so we can draw colored rectangles later
img = cv2.imread(image_filepath)


if img is None:
  print(f"Error: unable to load image from {image_filepath}")
  exit()

# load the pre trained Haar cascade XML file for frontal face detection
face_cascade = cv2.CascadeClassifier(
    cv2.data.haarcascades + "haarcascade_frontalface_default.xml"
)

# convert the color image to grayscale for face detection (this is now correct)
gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# detect faces in image
## detectMultiScale() detects object (faces) of different sizes
faces = face_cascade.detectMultiScale(
    gray_img,
    scaleFactor=1.1, # how much the size is reduced at each scale
    minNeighbors=5,  # Higher value = fewer false positives
    minSize=(30, 30) # minimum size of detected face
)

# 'faces' contains a list of rectangles:
# (x,y) -> top-left corner
# w -> width of the rectangle
# h -> height of the rectangle

# step5 draw rectangles around faces

# loop through all detected faces

for(x, y, w, h) in faces:
    # draw a rectangle around each detected face on the original color image
    # (255, 0, 0) --> blue color in BGR format
    # 2 -> thicknes of the rectangle
    cv2.rectangle(img, (x, y), (x + w, y + h), (255, 0, 0), 2)

# step 6 display the result
cv2_imshow(img)
```

```python
cv2.imwrite('detected_face.jpg', img)
```
