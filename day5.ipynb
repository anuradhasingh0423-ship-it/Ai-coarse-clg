```python
# fillna
# if datatype of col is numeric then we fill null values by mean/median of that col
# if datatype of col is obj then we fill it by mode (most repeated values) of that col
# if null values are more than 40% then we drop it
# df[col].fillna(df[col].mean(),inplace=True)  for numeric
# df[col].fillna(df[col].mode()[0],inplace=True)  for object col
```

```python
# create a df having info of your 5 family members with their id,name,age and gender
import pandas as pd

data = {
    "ID": [1, 2, 3, 4, 5],
    "Name": ["Father", "Mother", "Brother", "Sister", "Me"],
    "Age": [48, 45, 22, 18, 20],
    "Gender": ["Male", "Female", "Male", "Female", "Female"]
}

df = pd.DataFrame(data)
print(df)

```
