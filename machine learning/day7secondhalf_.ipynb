```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('Cars93.csv')
df.head()
```

```python
df = df[['Price','MPG.city']]
df.head()
```

```python
# create 3 clusters in Cars93 dataset using column Price and MPG.city
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# Prepare the features for clustering
X = df[['Price','MPG.city']]

# Feature scaling
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# Apply K-Means clustering
kmeans = KMeans(n_clusters=3, random_state=42)
df['Cluster'] = kmeans.fit_predict(X_scaled)
```

```python
# Plot the clusters
plt.figure()
plt.scatter(X['Price'], X['MPG.city'], c=df['Cluster'])
plt.xlabel("Price")
plt.ylabel("MPG.city")
plt.title("K-Means Clustering on Cars93 Dataset")
plt.show()
```

K-Means is an unsupervised learning algorithm used to group similar data points into clusters. In this task, the Cars93 dataset is clustered using two features: Price and MPG.city. Since K-Means is distance-based, the data is first standardized to ensure fair clustering. The algorithm then forms three clusters (K = 3) by assigning each car to the nearest centroid and updating the centroids iteratively until convergence. The resulting clusters represent groups of cars with similar price and city mileage characteristics.
