```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score, mean_squared_error
```

```python
df = pd.read_csv("Cars93.csv")
df.head()
```

```python
# check null values
print(df.isnull().sum())
```

```python
df = df.dropna()
df.head()
```

```python
print(df.isnull().sum())
```

```python
# fill numeric nulls with mean
num_cols = df.select_dtypes(include=['int64','float64']).columns
df[num_cols] = df[num_cols].fillna(df[num_cols].mean())
```

```python
# fill categorical nulls with mode
cat_cols = df.select_dtypes(include='object').columns
for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
X = df.drop("Price", axis=1)
y = df["Price"]
```

```python
#Divide categorical & numerical columns
cat = X.select_dtypes(include='object').columns
con = X.select_dtypes(exclude='object').columns
```

```python
#Create DataFrames of categorical & continuous
xcat = X[cat]
xcon = X[con]
```

```python
#Feature Encoding (Label Encoder)
le = LabelEncoder()

for col in xcat.columns:
    xcat[col] = le.fit_transform(xcat[col])
```

```python
#Feature Scaling (StandardScaler)
sc = StandardScaler()
xcon_scaled = sc.fit_transform(xcon)

xcon = pd.DataFrame(xcon_scaled, columns=xcon.columns)
xcon
```

```python
X_final = pd.concat([xcon, xcat], axis=1)
```

```python
xtrain, xtest, ytrain, ytest = train_test_split(
    X_final, y, test_size=0.2, random_state=42
)
```

```python
rf = RandomForestRegressor(
    n_estimators=10,
    random_state=42
)

rf.fit(xtrain, ytrain)
```

```python
# model evaluation
ypred = rf.predict(xtest)

r2 = r2_score(ytest, ypred)
rmse = np.sqrt(mean_squared_error(ytest, ypred))

print("R² Score:", r2)
print("RMSE:", rmse)
```
